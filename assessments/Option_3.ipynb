#**Abstract**

Nitrogen dioxide (NO₂) is one of the major urban air pollutants and has the potential to affect human health. Urban environments, particularly large cities like Beijing, experience elevated NO₂ levels due to factors such as heavy traffic and industrial activity, placing more people at risk. This investigation analysed air quality data from four different sites across Beijing to study the relationships between NO₂ readings at each site and to assess for health risks. Results showed there were strong positive correlations between NO₂ levels across the four sites, with some variation that could have been due to several factors at each site, such as windspeed or traffic volume. Additionally, the median NO₂ value at each site varied between 47.0 and 76.0 µg/m³. Both of these values are above the 2021 WHO Air Quality Guidelines (AQGs) for both short-term (daily) and long-term (annual) exposure, which are 25 and 10 µg/m³, respectively. The maximum values ranged between 262.0 and 300.0 µg/m³, considerably above these guidelines and posing a serious threat to public health. Only 6.09% of all NO₂ values were below the WHO daily maximum recommended exposure, meaning nearly 94% of the time people in the area may have been at increased risk of health issues due to elevated NO₂ concentration. Possibly more worryingly, only 0.0014% of NO₂ values were below the annual recommendations. Having unsafe NO₂ concentrations for over 99.99% of the year is a serious concern for public health around the four sites in Beijing.

#**Introduction**

Nitrogen dioxide (NO₂) tends to be released into the atmosphere through human activities such as industrial activity, fossil fuel burning, biomass burning, heavy traffic movement, electricity generation, and power plants (Jion et al., 2023). These sources are concentrated in greater capacity and numbers in urban areas, meaning cities across the globe experience higher volumes of NO₂ pollution. When combined with the greater population density of urban areas, this becomes a threat to public health due to the larger quantity of people exposed to poor air quality (Jion et al., 2023).
  
Several studies documented associations between day-to-day variations in NO₂ and variations in mortality, hospital admissions, and respiratory diseases. There is also evidence of correlation between chronic exposure to NO₂ and mortality and morbidity (Solange Costa et al., 2014). Beijing is one of the world's largest cities, and suffers from air quality challenges as a result. Supporting industry, transportation, and the energy needs of a population exceeding 20 million generates higher levels of NO₂ than are typically observed in most other regions of the world. This could potentially have a very serious impact on public health, as  NO₂ has been linked to several serious health issues. Nitrogen dioxide inflames the lining of the lungs and reduces the body's ability to fight infection (OpenAI, 2025). Vulnerable groups such as the elderly, people with chronic respiratory and/or cardiovascular conditions or young children are most at risk. However, all people could potentially be affected by short and long term conditions promoted by NO₂. Potential health issues range from increased susceptibility to respiratory infections (e.g. pneumonia) to permanent loss of lung function in children and higher risk of premature death from respiratory and cardiovascular causes (World Health Organization, 2021).

NO₂ also plays a role in the development of photochemical smog, a brownish-grey haze caused by the action of ultraviolet radiation on an atmosphere polluted with hydrocarbons and oxides of nitrogen (NO₂) (Sher, 1998). Photochemical smog can have adverse effects on human health as it contains toxic pollutants such as ozone and nitric acid (Sher, 1998). Furthermore, photochemical smog has an unpleasant odour and can block out sunlight, which can negatively affect the mental health of people consistently exposed to it.Bejing photochemical smog.avif
######Figure 1 shows the Beijing skyline immersed in photochemical smog. The haze was produced from sunlight interacting with oxides of nitrogen emitted from combustion engines interacting with hydrocarbons in the presence of ultraviolet radiation. The thickness of the pollution can be seen in this image, with objects in the background difficult to make out through the haze (Fuller, 2024).

Air quality has been an issue in China for a considerable length of time, with 2.6 million air pollution related deaths in 2005 alone. Beijing was named the smog capital of the world in 2008, causing issues for the 2008 Olympics, when concerns for the health of athletes forced the city of Beijing to improve its air quality using temporary measures (Fuller, 2024).

Extreme air pollution has more than just social impacts, with China losing over US\$38 billion annually (measured in the number of air-pollution related deaths and crop production losses). Crop production is significantly hindered by poor air quality, with China losing an annual total of 20 million tonnes of rice, wheat, maize, and soybeans (Anuszkiewicz et al., 2023). There is a further economic cost for attempts to remove pollutants from the atmosphere with US\$10 billion spent on temporarily improving conditions for the 2008 Olympics.

From 2008 to 2022, air pollution improved in the region by nearly 30%, yet remained significantly higher than the WHO Air Quality Guidelines (AQGs), meaning air pollution in Beijing is still placing public health at risk (Baiyu, 2022). Additionally, crop and plant damage along with the economic cost of lives, healthcare and mitigation strategies means the Chinese environment and financial system are also suffering due to poor air quality.

This investigation aims to study the nitrogen dioxide levels at four sites in Beijing (Nansanhuan, Yizhuang, Yongdingmennei, Qianmen) from January 2017 to December 2018. The purpose of this is to assess potential risk to public health during this time and to examine potential relationships between sites of interest, which might suggest common causes for NO₂ values.

````python
#Import modules.
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
````

````python
#Load data.
data_aq = pd.read_csv("data/beijing_17_18_aq.csv")
````

````python
#Preview data.
data_aq.head()
````

````python
#Check the shape, size and number of dimensions of data_aq.
print(data_aq.shape)
print(data_aq.size)
print(data_aq.ndim)
````

````python
#Convert 'utc_time' to datetime format and preview data types to check if it has worked. The datetime format will allow python to convert string timestamps to datetime for calculations. The utc_time row should show "datetime64[ns]".
data_aq['utc_time'] = pd.to_datetime(data_aq['utc_time'], format='mixed')
data_aq.dtypes
````

````python
#Set index to utc_time. This will mean all sub-dataframes will have the same index.
data_aq = data_aq.set_index('utc_time')
````

````python
#Create new dataframes for each station ID.
nansanhuan_aq_df = data_aq[data_aq['stationId'] == 'nansanhuan_aq']
yizhuang_aq_df = data_aq[data_aq['stationId'] == 'yizhuang_aq']
yongdingmennei_aq_df = data_aq[data_aq['stationId'] == 'yongdingmennei_aq']
qianmen_aq_df = data_aq[data_aq['stationId'] == 'qianmen_aq']
````

````python
#Rename NO2 columns to avoid confusion when merged.
nansanhuan_aq_df = nansanhuan_aq_df.rename(columns={"NO2": "NO2_nansanhuan"})
yizhuang_aq_df = yizhuang_aq_df.rename(columns={"NO2": "NO2_yizhuang"})
yongdingmennei_aq_df = yongdingmennei_aq_df.rename(columns={"NO2": "NO2_yongdingmennei"})
qianmen_aq_df = qianmen_aq_df.rename(columns={"NO2": "NO2_qianmen"})
````

````python
#Make each dataframe only include Station ID and NO2.
nansanhuan_aq_df = nansanhuan_aq_df[['stationId', 'NO2_nansanhuan']]
yizhuang_aq_df = yizhuang_aq_df[['stationId', 'NO2_yizhuang']]
yongdingmennei_aq_df = yongdingmennei_aq_df[['stationId', 'NO2_yongdingmennei']]
qianmen_aq_df = qianmen_aq_df[['stationId', 'NO2_qianmen']]

#Preview one of the dataframes.
yizhuang_aq_df.head()

````

````python
#Merge all the dataframes and rename stationID column to stationID_"site name".
merged_df = pd.merge(nansanhuan_aq_df, yizhuang_aq_df, left_index=True, right_index=True, how='outer', suffixes=('_nansanhuan', '_yizhuang'))
merged_df = pd.merge(merged_df, yongdingmennei_aq_df, left_index=True, right_index=True, how='outer', suffixes=('_yizhuang', '_yongdingmennei'))
merged_df = pd.merge(merged_df, qianmen_aq_df, left_index=True, right_index=True, how='outer', suffixes=('_yongdingmennei', '_qianmen'))

#The left/right_index = True section allows the dataframes to merge on the "utc_time" index.
#The how = 'outer' section means all timestamps from all dataframes are preserved.
#The suffixes=() section avoids duplicate column names (stationID) in the merged dataframe.

#Preview the merged dataframe.
merged_df.head()
````

````python
#Produce a correlation coefficient matrix.
correlation_matrix = merged_df[['NO2_nansanhuan', 'NO2_yizhuang', 'NO2_yongdingmennei', 'NO2_qianmen']].corr()
correlation_matrix
````

````python
#Define the figure size and axes. ChatGPT was used to "clean up" the original figure, for example rotating the labels on the x-axis. The original code (provided in the appendix) was input to ChatGPT along with this prompt:
#Clean up the x and y labels, remove "NO2_" from each of the labels, express "NO2" in the title with the "2" as a subscript (OpenAI, 2025).
fig, axes = plt.subplots(1, 2, figsize=(14, 6))

#Remove 'NO2_' from heatmap labels.
correlation_matrix = correlation_matrix.rename(
    columns=lambda x: x.replace("NO2_", ""),
    index=lambda x: x.replace("NO2_", "")
)

#Produce a heatmap on the first axis and give it a title.
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', ax=axes[0])

#Clean up heatmap axis labels.
axes[0].set_title('Figure 2 - Correlation Heatmap of NO$_2$ Values (µg/m³)', fontsize=14)
axes[0].set_xticklabels(axes[0].get_xticklabels(), rotation=45, ha='right', fontsize=10)
axes[0].set_yticklabels(axes[0].get_yticklabels(), rotation=0, fontsize=10)

#Extract column names for boxplot and remove 'NO2_'.
no2_columns = ['NO2_nansanhuan', 'NO2_yizhuang', 'NO2_yongdingmennei', 'NO2_qianmen']
clean_labels = [label.replace('NO2_', '') for label in no2_columns]

#Produce a boxplot on the second axis and give it a title.
sns.boxplot(data=merged_df[no2_columns], ax=axes[1])

#Clean up boxplot axis labels
axes[1].set_title(r'Figure 3 - Boxplot of NO$_2$ Values', fontsize=14)  #NO₂ using LaTeX.
axes[1].set_xticks(range(len(clean_labels)))  #Set positions (0, 1, 2, 3).
axes[1].set_xticklabels(clean_labels, rotation=30, ha='right', fontsize=10)
axes[1].set_ylabel(r'NO$_2$ Concentration (µg/m³)', fontsize=12)  #NO₂ on the y-axis too.

#Show the plots.
plt.tight_layout()
plt.show()
````

````python
from scipy.stats import kruskal, mannwhitneyu
from itertools import combinations

# Define your dataframe: merged_df
columns = ['NO2_nansanhuan', 'NO2_yizhuang', 'NO2_yongdingmennei', 'NO2_qianmen']
data = {col: merged_df[col].dropna() for col in columns}

# Step 1: Kruskal-Wallis H-test (global test)
stat, p_value = kruskal(*data.values())

print("Kruskal-Wallis H-test:")
print(f"Statistic = {stat:.4f}, p-value = {p_value:.4f}")
if p_value < 0.05:
    print("→ Significant differences detected among the medians. Proceeding to pairwise comparisons.\n")
else:
    print("→ No significant difference in medians. No need for pairwise testing.\n")

# Step 2: Pairwise Mann-Whitney U tests with Bonferroni correction
alpha = 0.05
comparisons = list(combinations(columns, 2))
corrected_alpha = alpha / len(comparisons)

print(f"Pairwise Mann-Whitney U tests (Bonferroni-corrected α = {corrected_alpha:.4f}):")
for col1, col2 in comparisons:
    stat, p = mannwhitneyu(data[col1], data[col2], alternative='two-sided')
    significant = p < corrected_alpha
    result = "Significant" if significant else "Not Significant"
    print(f"{col1} vs {col2}: U = {stat:.2f}, p = {p:.4f} → {result}")
````

#**Further research**

The relationship between ground-level ozone (O₃) and NO₂ plays a major role in photochemical smog formation. Investigating relative O₃ and NO₂ values across Beijing would provide useful insight with regard to estimating the risk of photochemical smog development (Sher, 1998).

A major stage in photochemical smog formation is the conversion of NO₂ to O₃. Analysing NO₂ and O₃ over time may show when this process occurs, indicating when photochemical smog is likely to form. The full process of photochemical smog formation is shown in Table 1 (Thomas A Hecht and Seinfeld, 1972).Screenshot 2025-05-11 132313.png

This process can be simplified into four main stages:

1. Formation of NO₂ - NO from vehicle exhausts interacts with atmospheric O₂ to form NO₂.
2. Photodissociation of NO₂ - NO₂ absorbs ultraviolet (UV) light and breaks down into NO + O.  
3. Ozone formation - The free O atom reacts with O₂ to form O₃.  
4. Role of Volatile Organic Compounds (VOCs) - VOCs (from fuels, paints, industrial processes, etc.) react with NO to form peroxy radicals (RO₂•), which inhibit the breakdown of ozone and help accumulate more O₃ and other harmful oxidants like peroxyacetyl nitrate (PAN). This causes photochemical smog to occur (OpenAI, 2025).

This process is most likely to occur in sunny, urban environments with high vehicle emissions. Beijing fits this profile.

As seen in steps 1-3, NO₂ is converted to O₃ during this process. This means a negative correlation is expected between these variables. Due to this the period where NO₂ concentration decreases and O₃ increases in a high UV environment will be considered as the most high risk time in terms of photochemical smog development.

Additionally, O₃ is one of the most toxic atmospheric pollutants, with the potential to interact with other pollutants and damage human health. This means studying its temporal patterns will benefit public health investigations.

````python
#Import modules.
import matplotlib.animation as animation
import matplotlib.dates as mdates
from IPython.display import HTML

#Calculate mean NO2 and O3 values across stations.
mean_no2 = merged_df_O3[['NO2_nansanhuan', 'NO2_yizhuang', 'NO2_yongdingmennei', 'NO2_qianmen']].mean(axis=1)  # Mean across columns
mean_o3 = merged_df_O3[['O3_nansanhuan', 'O3_yizhuang', 'O3_yongdingmennei', 'O3_qianmen']].mean(axis=1)  # Mean across columns

#Prepare data.
monthly_mean_no2 = mean_no2.resample('ME').mean()
monthly_mean_o3 = mean_o3.resample('ME').mean()

#Convert index to values for animation.
dates = monthly_mean_no2.index
no2_vals = monthly_mean_no2.values
o3_vals  = monthly_mean_o3.values

#Set up the figure and axes.
fig, ax = plt.subplots(figsize=(14, 6))
ax.set_xlim(dates[0], dates[-1])
ax.set_ylim(0, max(max(no2_vals), max(o3_vals)) * 1.1)
ax.set_title('Figure 13 - Animated Monthly Average NO₂ and O₃ Concentrations')
ax.set_xlabel('Month')
ax.set_ylabel('Concentration (µg/m³)')
ax.xaxis.set_major_formatter(mdates.DateFormatter('%b %Y'))
plt.xticks(rotation=45)

#Initialize empty lines.
line_no2, = ax.plot([], [], label='Avg Monthly NO₂', color='tab:blue')
line_o3,  = ax.plot([], [], label='Avg Monthly O₃', color='tab:orange')
ax.legend()
ax.grid(True)

#Animation function.
def animate(i):
    line_no2.set_data(dates[:i], no2_vals[:i])
    line_o3.set_data(dates[:i], o3_vals[:i])
    return line_no2, line_o3

#Create the animation.
ani = animation.FuncAnimation(fig, animate, frames=len(dates)+1, interval=200, blit=True)

#Display animation in Jupyter (OpenAI, 2025).
HTML(ani.to_jshtml())
````
